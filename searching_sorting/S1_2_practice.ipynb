## Linear Search

```python
def Linear_Search(arr, target):
    size = len(arr)
    for index in range(size):
        if arr[index] == target:
            return f'{target} is found at {index}'
    return -1
```

```python
lst = [23,43,65,78,342,54,34,56,'g']
print(Linear_Search(lst, 'g'))
```

## Binary Search

```python
def Binary_Search(arr, target):
    size = len(arr)
    arr.sort()
    low = 0
    high = size - 1

    while(low <= high):
        mid = (low + high) // 2   #we used floor division
        if arr[mid] == target:
            return f'{target} is found at {mid}'
        elif arr[mid] < target:
            low = mid + 1
        elif arr[mid] > target:
            high = mid - 1
    return f"{target} is not present in the list"    
```

```python
lst = [4,6,7,8,9,2,33,555,66,33]
print(Binary_Search(lst, 555))
```

## Bubble Sort

```python
def bubble_sort(arr):
    n = len(arr)
    phase = n
    for j in range(phase):
        for i in range(n-1):
            if arr[i] > arr[i+1]:
                c = arr[i+1]
                arr[i+1] = arr[i]
                arr[i] = c
            else:
                continue
            print(arr)
            phase-= 1
    return 0

lst = [ 90,2,14,6,54,3,23]
bubble_sort(lst)
```
