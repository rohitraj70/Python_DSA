#### Type Inference
Type inference means Python can determine the type of a variable automatically based on the value you assign to it.

```python
x = 9
print(type(x))
x = "ji"
print(type(x))
```

#### Variables

variables can be defined as the container that contains a value in it ,
such as in x = 8 , x is the container which keeps 8 in it .

```python
a = 9
print(a)  # this will give the value of a
print("a")  # this will give the string "a"
```

# Conversion of types , type conversion 

```python
age = 34
str_age = str(age)   #type converted
print(type(str_age))
# more are type conversions in python
int_age = int(str_age)  # type converted
print(type(int_age))  
#int -> float is possible but reverse is not possible if the float has decimal values
l = 1
b = bool(l)  # True = 1 and False = 0
print(b)
#bool to int is possible but reverse is not possible if the int is not 0 or 1
```
