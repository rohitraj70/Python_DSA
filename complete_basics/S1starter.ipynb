# Why we need to install or have virtual environment ?

Suppose if we have multiple projects and each of them require different versions of some requied dependencies . So for each project we create virtual environment for each projects separately . 

## How to create virtual environment ?

```python
sudo apt update                  # Update package list
sudo apt install python3-venv    # Install venv support
python3 -m venv .venv             # Create environment
source .venv/bin/activate         # Activate environment
deactivate
```
