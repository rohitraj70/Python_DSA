##### Standard Library Overview
Python's Standard Library is a vast collection of modules and packages that come bundled with Python, providing a wide range of functionalities out of the box. Here's an overview of some of the most commonly used modules and packages in the Python Standard Library.

```python
import array
arr=array.array('i',[1,2,3,4])
print(arr)
```

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
## random 

import random
print(random.randint(1,10))
print(random.choice(['apple','banana','cherry']))
```

```python
### File And Directory Access

import os
print(os.getcwd())  #print current working directory
```

```python
os.mkdir('test_dir')   #create a new directory
#to create a new file in the directory
with open('test_file.txt', 'w') as f:  #this will create a new file named test_file.txt in the current working directory and open it in write mode
    f.write('Hello, World!')
os.chdir('test_dir')  #change current working directory to test_dir
```

```python
from pathlib import Path
print(Path.cwd())  #print current working directory using pathlib
```

```python
## High level operations on files and collection of files
print(os.getcwd())

import shutil
# os.chdir('..')  #change current working directory to parent directory
shutil.copyfile('source.txt','destination.txt')   #destination.txt will be created in the current working directory and the contents of source.txt will be copied to it
```

```python
os.chdir('..')  #change current working directory to parent directory
print(os.getcwd())
```

```python
## Data Serialization
import json
data={'name':'Krish','age':25}  #dictionary data type, converting to json format via json.dumps() method

json_str=json.dumps(data)    #api give a string representation of the data in json format
print(json_str)
print(type(json_str))

parsed_data=json.loads(json_str)    #parsing the json string back to a Python dictionary to access the data in a structured way 
print(parsed_data)
print(type(parsed_data))
```

```python
## csv

import csv

with open('example.csv',mode='w',newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Krish',32])

with open('example.csv',mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
## datetime
from datetime import datetime,timedelta

now=datetime.now()
print(now)

yesterday=now-timedelta(days=1)

print(yesterday)
```

```python
## time
import time
print(time.time())
time.sleep(2)
print(time.time())
```

```python
## Regular expresiion
import re

pattern=r'\d+'   #check for one or more digits in the string
text='There are 123 apples 456'
match=re.search(pattern,text)
print(match.group())
```
