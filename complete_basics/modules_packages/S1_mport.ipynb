##### Importing Modules in Python: Modules and Packages
In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math   # math is a package that provides mathematical functions and constants. It is part of the Python Standard Library and can be used to perform various mathematical operations, such as calculating square roots, trigonometric functions, logarithms, and more.
math.sqrt(16)

# the difference between a library, module and package in Python is as follows:
# 1. Library: A library is a collection of modules and packages that provide specific functionality. It is a broader term that encompasses multiple modules and packages. For example, the Python Standard Library is a collection of modules and packages that provide various functionalities, such as file I/O, networking, and data manipulation.
# 2. Module: A module is a single file that contains Python code, which can include functions, classes, and variables. Modules are used to organize code into reusable components. For example, the `math` module is a single file that contains mathematical functions and constants.
# 3. Package: A package is a directory of modules. It is a way to organize related modules together. For example, the `numpy` package is a collection of modules that provide numerical computing functionality.    
```

```python
from math import sqrt,pi   #to import specific functions or constants from a module, you can use the `from` keyword followed by the module name and the names of the functions or constants you want to import. This allows you to use those functions or constants directly without having to prefix them with the module name.
print(sqrt(16))
print(sqrt(25))
print(pi)
```

```python
import numpy as np    #np is a commonly used alias for the NumPy package, which is a powerful library for numerical computing in Python. It provides support for large, multi-dimensional arrays and matrices, along with a wide range of mathematical functions to operate on these arrays. By importing NumPy as np, you can use the np prefix to access its functions and classes, making your code more concise and readable.

np.array([1,2,3,4])
```

```python
from math import *    # The `from math import *` statement imports all functions and constants from the `math` module into the current namespace. This means you can use any function or constant from the `math` module without needing to prefix it with `math.`. However, using this approach can lead to potential naming conflicts if there are functions or variables with the same name in your code or other imported modules. It is generally recommended to import only the specific functions or constants you need, or to use an alias for the module to avoid such conflicts.
print(sqrt(16))
print(pi)
```

```python
from package.maths import addition
addition(2,3)
```

```python
from package import maths
maths.add(2,3)
```

```python
from package.maths import add as addi
addi(2,3)
```

```python
from package.maths import *
add(2,3)
```

```python
#subpackage is a package that is contained within another package. It allows for further organization and structuring of code within a larger package. Subpackages can have their own modules and can be imported and used just like regular packages. They provide a way to create a hierarchy of packages, allowing for better organization and separation of functionality within a larger codebase.
from package.subpackage import mult
mult.multiply(2,3)
```

```python
import package.subpackage.mult as m
m.multiply(2,3)
```

```python
import package.subpackage.mult 
package.subpackage.mult.multiply(2,3)
```
